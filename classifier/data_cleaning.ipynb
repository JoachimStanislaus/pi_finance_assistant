```python
import pandas as pd

df = pd.read_csv("Expense_desc_cat_data.csv")
```

```python
print(df["category"].unique())
```

```python
print(df["category"].value_counts())
```

```python
# Remove rows with following categories
df = df[~df["category"].isin(["Car"])]
```

```python
print(sorted(df["category"].dropna().unique()))
```

```python
df["category"] = df["category"].replace({
    "Entertainment / Social": "Social",
})
```

```python
df = df.dropna(subset=["category"])
```

```python
df.to_csv("datasets/expenses.csv", index=False)
```
